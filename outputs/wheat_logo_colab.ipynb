# Wheat Logo Generator
Run all cells (Runtime > Run all). Generates the SVG, previews it, exports a PNG, and downloads both.

```python
# Wheat icon logo generator (Colab-compatible)

KERNEL = '<path d="M0 -45 C 15 -26 17 20 0 46 C -17 20 -15 -26 0 -45 Z"/>'

ICON = f'''<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 512 512" width="512" height="512">
  <defs>
    <linearGradient id="bg" x1="0" y1="0" x2="1" y2="1">
      <stop offset="0" stop-color="#2f7d46"/>
      <stop offset="1" stop-color="#1c5530"/>
    </linearGradient>
    <linearGradient id="gold" x1="0" y1="0" x2="0" y2="1">
      <stop offset="0" stop-color="#f2e3b3"/>
      <stop offset="1" stop-color="#d9bd7a"/>
    </linearGradient>
  </defs>
  <!-- tile -->
  <rect x="16" y="16" width="480" height="480" rx="112" fill="url(#bg)"/>
  <rect x="16" y="16" width="480" height="480" rx="112" fill="none"
        stroke="#7fce9a" stroke-opacity="0.35" stroke-width="6"/>
  <!-- stem -->
  <path d="M256 175 V414" stroke="#dcc084" stroke-width="16" stroke-linecap="round" fill="none"/>
  <!-- wheat kernels -->
  <g fill="url(#gold)">
    <g transform="translate(256,142)">{KERNEL}</g>
    <g transform="translate(224,218) rotate(-38)">{KERNEL}</g>
    <g transform="translate(288,218) rotate(38)">{KERNEL}</g>
    <g transform="translate(220,282) rotate(-38)">{KERNEL}</g>
    <g transform="translate(292,282) rotate(38)">{KERNEL}</g>
    <g transform="translate(222,346) rotate(-38)">{KERNEL}</g>
    <g transform="translate(290,346) rotate(38)">{KERNEL}</g>
  </g>
</svg>'''

with open('wheat_icon.svg', 'w') as f:
    f.write(ICON)
print('Saved wheat_icon.svg')
```

```python
# Preview inline
from IPython.display import SVG, display
display(SVG('wheat_icon.svg'))
```

```python
# Export PNG (512px)
!pip install cairosvg -q
import cairosvg
cairosvg.svg2png(url='wheat_icon.svg', write_to='wheat_icon.png', output_width=512)
print('PNG saved')
```

```python
# Download files
from google.colab import files
files.download('wheat_icon.svg')
files.download('wheat_icon.png')
```

```python
# Favicon (.ico with 16/32/48/64 px) + web icons
!pip install cairosvg pillow -q
import cairosvg, io
from PIL import Image

sizes = [16, 32, 48, 64]
imgs = []
for s in sizes:
    png = cairosvg.svg2png(url='wheat_icon.svg', output_width=s, output_height=s)
    imgs.append(Image.open(io.BytesIO(png)).convert('RGBA'))
imgs[-1].save('favicon.ico', format='ICO', sizes=[(s, s) for s in sizes], append_images=imgs[:-1])
cairosvg.svg2png(url='wheat_icon.svg', write_to='apple-touch-icon.png', output_width=180)

from google.colab import files
files.download('favicon.ico')
files.download('apple-touch-icon.png')
```
